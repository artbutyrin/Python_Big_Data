https://towardsdatascience.com/do-you-use-apply-in-pandas-there-is-a-600x-faster-way-d2497facfa66

```python
df['annual_income_num'] = df['Annual_Income'].map(lambda val: 1 if val == '<=50K' else 2)
df
```

```python
def is_eligible(row):
    if row['Education'] in ('Bachelors', 'Masters') and row['age_of_person'] < 26:
      return True
    return False

df['is_scholarship_eligible'] = df.apply(is_eligible, axis=1)
df[df['is_scholarship_eligible']]
```
