Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
class Cylinder(figure3D):
  def __init__(self, r, h):
    self.r = r
    self.h = h

  def get_volume(self):
    pi = 3.14
    return pi * self.r ** 2 * self.h

  def __add__(self, other):
    return self.get_volume() + other.get_volume()

class Cube:
  def __init__ (self, a):
    self.a = a

  def get_volume(self):
    return self.a ** 3

  def __add__(self, other):
    return self.get_volume() + other.get_volume()

c1 = Cylinder(2, 4)
c2 = Cube(2)

print(c1 + c2)
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
  def __init__(self, id, balance=0):
    self.__id = id
    self.__balance = balance

  @property
  def balance(self):
      return self.__balance

  @balance.setter
  def balance(self, value):
    if value < 0:
      raise ValueError("Баланс не може бути від'ємним")
    self.__balance = value

  def deposit(self, amount):
    self.balance = self.__balance + amount

  def withdraw(self, amount):
    if self.__balance < amount:
      raise ValueError("Недостатньо коштів")
    self.__balance -= amount
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
  def __init__(self, name, price):
    self.name = name
    self.price = price

  def __repr__(self):
    return f"{self.name}: {self.price}"

class ShoppingCart:
  def __init__(self):
    self._items = []

  def add(self, item):
    self._items.append(item)

  def remove(self, item):
    if item in self._items:
      self._items.remove(item)

  def show(self):
      return self._items

cart = ShoppingCart()
cart.add(Item("Мишка", 4))
cart.add(Item("клавіатура", 2))
cart.remove(Item("Мишка", 4))
print(cart.show())
```

```python
class Vehicle:
  def __init__(self, brand, production_age):
    self.brand = brand
    self.production_age = production_age

class Engine:
  def __init__(self, type_, size):
    self.type_ = type_
    self.size = size

class Car(Vehicle, Engine):
  def __init__(self, brand, production_age, type_, size):
    Vehicle.__init__(self, brand, production_age)
    Engine.__init__(self, type_, size)

car = Car("Toyota", 2020, "petrol", 2.0)
print(car.brand, car.type_, car.size)
print(Car.mro())
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekDay:
  DAYS = ["Понеділок", "вівторок", "середа", "четвер", "п'ятниця", "субота", "неділя"]

  def __iter__(self):
    self._iterator = 0
    return self

  def __next__(self):
    if self._iterator >= len(self.DAYS):
      raise StopIteration
    name = self.DAYS[self._iterator]
    self._iterator += 1
    return name

weekdaays = WeekDay()
for day in weekdaays:
  print(day)
```
